<font color="red" size="6"> Filtering </font>

<font color="red" size="6"> GroupBy Analysis </font>

<font color="yellow" size="5">Gender wise average math score </font>

```python
math_average=data.groupby('gender')['math score'].mean()
```

<font color="yellow" size="5"> Gender wise average reading score </font>

```python
read_average=data.groupby('gender')['reading score'].mean()
```

<font color="yellow" size="5"> Lunch type wise average writing score </font>

```python
write_average=data.groupby('lunch')['writing score'].mean()
```

<font color="yellow" size="5"> Parental education wise average math score </font>

```python
math_parent_average=data.groupby('parental level of education')['math score'].mean()
```

<font color="red" size="6"> Histogram Chart Student Math Score </font>

```python
plt.hist(data['math score'],bins=13)
plt.title("Student Math Score ")
plt.xlabel("Math Score")
plt.ylabel("Frequency")
plt.savefig("histogram_chart.png")
plt.show()
```

<font color="red" size="6"> Bar Chart Male and Female average Math Score </font>

```python
plt.bar(math_average.index,math_average.values)
plt.title("Average Male and Female Math Score")
plt.xlabel("Gender")
plt.ylabel("Math Score")
plt.savefig("bar_chart.png")
plt.show()
```
